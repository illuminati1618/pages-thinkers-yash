    <h2><span class="section-icon">📋</span>Phase 1: Building the Player Class</h2>
    <p>Let's start with the simplest class - a Player that has a name and symbol.</p>

    <h3>Anatomy of a Class</h3>
    <div class="code-preview">

class Player:                    ← Class definition
    def __init__(self, name, symbol):  ← Constructor method
        self.name = name         ← Attribute assignment
        self.symbol = symbol     ← Attribute assignment

</div>

    <h3>Understanding the Parts</h3>
    <ul>
        <li><strong><code>class Player:</code></strong> - Defines a new class called Player</li>
        <li><strong><code>__init__()</code></strong> - Special method that runs when creating a new object</li>
        <li><strong><code>self</code></strong> - Refers to the specific object being created</li>
        <li><strong><code>self.name</code></strong> - Creates an attribute called 'name' for this object</li>
    </ul>

    <div class="info-box warning-box">
        <h4>⚠️ Important</h4>
        <p><code>self</code> always comes first in method definitions, but you don't pass it when creating objects!</p>
    </div>

```python
# 🔧 Mini-Challenge 1: Create Your First Class

class Player:
    def __init__(self, name, symbol):
        self.name = name
        self.symbol = symbol

# Now let's create some player objects!
print("Creating player objects...")

player1 = Player("Alice", "X")
player2 = Player("Bob", "O")

print(f"Player 1: {player1.name} plays as '{player1.symbol}'")
print(f"Player 2: {player2.name} plays as '{player2.symbol}'")
print()

# Let's explore what we created
print("Object details:")
print(f"player1 is of type: {type(player1)}")
print(f"player1.name contains: {repr(player1.name)}")
print(f"player1.symbol contains: {repr(player1.symbol)}")
```

<div class="section">
    <h2><span class="section-icon">🎨</span>Phase 2: Building the Board Class</h2>
    <p>Now let's create a more complex class that handles the game board. This class will have both <strong>data</strong> (the grid) and <strong>behavior</strong> (methods to display and manipulate the board).</p>

    <h3>Methods vs Functions</h3>
    <div class="code-preview">

# Function (what we used before)
def print_board(board):
    # code here

# Method (belongs to a class)
class Board:
    def display(self):         ← Method (note the 'self')
        # code here

</div>

    <h3>Key Differences:</h3>
    <ul>
        <li><strong>Functions</strong> are independent - <code>print_board(my_board)</code></li>
        <li><strong>Methods</strong> belong to objects - <code>my_board.display()</code></li>
        <li>Methods automatically know which object they're working with (via <code>self</code>)</li>
    </ul>
</div>

```python
# 🔧 Mini-Challenge 2: Build the Board Class

class Board:
    def __init__(self):
        """Initialize an empty 3x3 board"""
        self.grid = [" "] * 9
    
    def display(self):
        """Show the current board state"""
        print("\n")
        print(" " + self.grid[0] + " | " + self.grid[1] + " | " + self.grid[2])
        print("---+---+---")
        print(" " + self.grid[3] + " | " + self.grid[4] + " | " + self.grid[5])
        print("---+---+---")
        print(" " + self.grid[6] + " | " + self.grid[7] + " | " + self.grid[8])
        print("\n")
    
    def display_reference(self):
        """Show position numbers for reference"""
        reference = ["1", "2", "3", "4", "5", "6", "7", "8", "9"]
        print("Board positions:\n")
        print("\n")
        print(" " + reference[0] + " | " + reference[1] + " | " + reference[2])
        print("---+---+---")
        print(" " + reference[3] + " | " + reference[4] + " | " + reference[5])
        print("---+---+---")
        print(" " + reference[6] + " | " + reference[7] + " | " + reference[8])
        print("\n")

# Test our Board class!
print("Creating a board object...")
board = Board()

print("Empty board:")
board.display()

print("Reference board:")
board.display_reference()

# Let's manually place some symbols to test
board.grid[0] = "X"  # Top-left
board.grid[4] = "O"  # Center
board.grid[8] = "X"  # Bottom-right

print("Board with some moves:")
board.display()
```

<div class="section">
    <h3>🎯 Adding More Board Methods</h3>
    <p>Our Board class needs more functionality. Let's add methods for making moves and checking game state.</p>
    
    <p>Each method should have a single, clear purpose:</p>
    
    <div class="code-preview">

make_move(position, symbol)  → Try to place a symbol
is_full()                   → Check if board is full
check_winner(symbol)        → Check if someone won

</div>
</div>

```python
# 🔧 Mini-Challenge 3: Complete Board Class with Game Logic

class Board:
    def __init__(self):
        """Initialize an empty 3x3 board"""
        self.grid = [" "] * 9
    
    def display(self):
        """Show the current board state"""
        print("\n")
        print(" " + self.grid[0] + " | " + self.grid[1] + " | " + self.grid[2])
        print("---+---+---")
        print(" " + self.grid[3] + " | " + self.grid[4] + " | " + self.grid[5])
        print("---+---+---")
        print(" " + self.grid[6] + " | " + self.grid[7] + " | " + self.grid[8])
        print("\n")
    
    def make_move(self, position, symbol):
        """Try to place a symbol at the given position (1-9)"""
        index = position - 1  # Convert to 0-8 index
        
        if index < 0 or index > 8:
            print("Invalid position. Choose a number between 1 and 9.")
            return False
        
        if self.grid[index] != " ":
            print("That spot is already taken. Try again.")
            return False
        
        self.grid[index] = symbol
        return True
    
    def is_full(self):
        """Check if the board is completely full"""
        return " " not in self.grid
    
    def check_winner(self, symbol):
        """Check if the given symbol has won the game"""
        win_combinations = [
            [0, 1, 2], [3, 4, 5], [6, 7, 8],  # Rows
            [0, 3, 6], [1, 4, 7], [2, 5, 8],  # Columns
            [0, 4, 8], [2, 4, 6]              # Diagonals
        ]
        
        for combo in win_combinations:
            if (self.grid[combo[0]] == symbol and
                self.grid[combo[1]] == symbol and
                self.grid[combo[2]] == symbol):
                return True
        return False

# Test all the methods!
print("Testing Board methods...")
board = Board()

print("Initial state:")
board.display()

print("Making moves:")
print("X at position 1:", board.make_move(1, "X"))
board.display()

print("O at position 1 (should fail):", board.make_move(1, "O"))
print("O at position 5:", board.make_move(5, "O"))
board.display()

print("Is board full?", board.is_full())
print("Does X win?", board.check_winner("X"))
print("Does O win?", board.check_winner("O"))
```

<div class="section">
    <h2><span class="section-icon">🎮</span>Phase 3: Building the Game Controller</h2>
    <p>Now we need a class that manages the entire game. This class will:</p>

    <div class="code-block">

TicTacToe Class Responsibilities:
┌─────────────────────────────┐
│ • Hold Player objects       │
│ • Hold Board object         │
│ • Manage whose turn it is   │
│ • Run the main game loop    │
│ • Handle input/output       │
│ • Determine game end        │
└─────────────────────────────┘

</div>

    <h3>Object Composition</h3>
    <p>Our TicTacToe class will contain other objects:</p>
    <div class="code-preview">

TicTacToe
├── board (Board object)
├── players (list of Player objects)
└── current_player (reference to a Player object)

</div>
    
    <p>This is called <strong>composition</strong> - building complex objects from simpler ones!</p>
</div>

```python
# 🔧 Mini-Challenge 4: Build the Game Controller

class TicTacToe:
    def __init__(self, player1, player2):
        """Initialize the game with two players"""
        self.board = Board()           # Create a Board object
        self.players = [player1, player2]  # Store Player objects
        self.current_player = player1  # Start with first player
    
    def switch_player(self):
        """Switch between player 1 and player 2"""
        if self.current_player == self.players[0]:
            self.current_player = self.players[1]
        else:
            self.current_player = self.players[0]
    
    def play(self):
        """Run the main game loop"""
        print("Welcome to Tic-Tac-Toe!")
        print(f"{self.players[0].name} is '{self.players[0].symbol}'")
        print(f"{self.players[1].name} is '{self.players[1].symbol}'")
        print("Players take turns choosing a position (1–9).\n")
        
        self.board.display_reference()
        self.board.display()
        
        # Main game loop
        while True:
            try:
                # Get player input
                move = int(input(f"{self.current_player.name} ({self.current_player.symbol}), enter your move (1-9): "))
            except ValueError:
                print("Invalid input. Please enter a number from 1 to 9.")
                continue
            
            # Try to make the move
            if not self.board.make_move(move, self.current_player.symbol):
                continue  # Move failed, try again
            
            # Show updated board
            self.board.display()
            
            # Check for winner
            if self.board.check_winner(self.current_player.symbol):
                print(f"{self.current_player.name} ({self.current_player.symbol}) wins!")
                break
            
            # Check for tie
            if self.board.is_full():
                print("It's a tie!")
                break
            
            # Switch to other player
            self.switch_player()

# Test the game setup (without full gameplay)
print("Setting up the game...")

# Create players
player1 = Player("Alice", "X")
player2 = Player("Bob", "O")

# Create game
game = TicTacToe(player1, player2)

print(f"Game created with {game.players[0].name} and {game.players[1].name}")
print(f"Current player: {game.current_player.name}")

# Test player switching
print("\nTesting player switching:")
print(f"Before switch: {game.current_player.name}")
game.switch_player()
print(f"After switch: {game.current_player.name}")
game.switch_player()
print(f"After second switch: {game.current_player.name}")
```

<div class="section">
    <h2><span class="section-icon">🏆</span>Phase 4: The Complete Object-Oriented Game</h2>
    <p>Now let's put all our classes together into the final game! Here's our complete object-oriented solution:</p>

    <h3>Object Interaction Diagram</h3>
    <div class="code-block">

User Input
    ↓
TicTacToe.play()
    ↓
Board.make_move() ← Uses current Player.symbol
    ↓
Board.check_winner() ← Checks for Player.symbol
    ↓
TicTacToe.switch_player() ← Changes current Player
    ↓
Loop back or End

</div>
</div>

```python
# 🎮 The Complete Object-Oriented Tic-Tac-Toe Game!

class Player:
    def __init__(self, name, symbol):
        self.name = name
        self.symbol = symbol


class Board:
    def __init__(self):
        self.grid = [" "] * 9

    def display(self):
        print("\n")
        print(" " + self.grid[0] + " | " + self.grid[1] + " | " + self.grid[2])
        print("---+---+---")
        print(" " + self.grid[3] + " | " + self.grid[4] + " | " + self.grid[5])
        print("---+---+---")
        print(" " + self.grid[6] + " | " + self.grid[7] + " | " + self.grid[8])
        print("\n")

    def display_reference(self):
        reference = ["1", "2", "3", "4", "5", "6", "7", "8", "9"]
        print("Board positions:\n")
        print("\n")
        print(" " + reference[0] + " | " + reference[1] + " | " + reference[2])
        print("---+---+---")
        print(" " + reference[3] + " | " + reference[4] + " | " + reference[5])
        print("---+---+---")
        print(" " + reference[6] + " | " + reference[7] + " | " + reference[8])
        print("\n")

    def is_full(self):
        return " " not in self.grid

    def make_move(self, position, symbol):
        index = position - 1
        if index < 0 or index > 8:
            print("Invalid position. Choose a number between 1 and 9.")
            return False
        if self.grid[index] != " ":
            print("That spot is already taken. Try again.")
            return False
        self.grid[index] = symbol
        return True

    def check_winner(self, symbol):
        win_combinations = [
            [0, 1, 2], [3, 4, 5], [6, 7, 8],  # Rows
            [0, 3, 6], [1, 4, 7], [2, 5, 8],  # Columns
            [0, 4, 8], [2, 4, 6]              # Diagonals
        ]
        for combo in win_combinations:
            if (self.grid[combo[0]] == symbol and
                self.grid[combo[1]] == symbol and
                self.grid[combo[2]] == symbol):
                return True
        return False


class TicTacToe:
    def __init__(self, player1, player2):
        self.board = Board()
        self.players = [player1, player2]
        self.current_player = player1

    def switch_player(self):
        self.current_player = (
            self.players[1] if self.current_player == self.players[0] else self.players[0]
        )

    def play(self):
        print("Welcome to Tic-Tac-Toe!")
        print(f"{self.players[0].name} is '{self.players[0].symbol}'")
        print(f"{self.players[1].name} is '{self.players[1].symbol}'")
        print("Players take turns choosing a position (1–9).\n")

        self.board.display_reference()
        self.board.display()

        while True:
            try:
                move = int(input(f"{self.current_player.name} ({self.current_player.symbol}), enter your move (1-9): "))
            except ValueError:
                print("Invalid input. Please enter a number from 1 to 9.")
                continue

            if not self.board.make_move(move, self.current_player.symbol):
                continue

            self.board.display()

            if self.board.check_winner(self.current_player.symbol):
                print(f"{self.current_player.name} ({self.current_player.symbol}) wins!")
                break

            if self.board.is_full():
                print("It's a tie!")
                break

            self.switch_player()


# Create and run the game!
if __name__ == "__main__":
    player1 = Player("Player 1", "X")
    player2 = Player("Player 2", "O")
    game = TicTacToe(player1, player2)
    game.play()
```
