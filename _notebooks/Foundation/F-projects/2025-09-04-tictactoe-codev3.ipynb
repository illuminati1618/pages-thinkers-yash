### 3.1 Board Initialization and Basic Structure

````python
# Step 1: Create the basic Board class with initialization
class Board:
    def __init__(self):
        """Initialize an empty 3x3 board"""
        # We use a list of 9 elements to represent our 3x3 grid
        # Positions 0-8 map to the board like this:
        # 0 | 1 | 2
        # ---------
        # 3 | 4 | 5  
        # ---------
        # 6 | 7 | 8
        self.grid = [" "] * 9  # 9 empty spaces
        print("New board created!")
        print(f"Grid representation: {self.grid}")

# Test the board creation
test_board = Board()
print(f"\nBoard has {len(test_board.grid)} positions")
print(f"All positions empty? {all(pos == ' ' for pos in test_board.grid)}")
````

### 3.2 Display Methods - Making the Board Visual

````python
# Step 2: Add display methods to our Board class
class Board:
    def __init__(self):
        self.grid = [" "] * 9

    def display(self):
        """Display the current game state"""
        print("\nCurrent Board:")
        print(" " + self.grid[0] + " | " + self.grid[1] + " | " + self.grid[2])
        print("---+---+---")
        print(" " + self.grid[3] + " | " + self.grid[4] + " | " + self.grid[5])
        print("---+---+---")
        print(" " + self.grid[6] + " | " + self.grid[7] + " | " + self.grid[8])
        print()

    def display_reference(self):
        """Show position numbers to help players"""
        print("\nPosition Reference (enter these numbers):")
        print(" 1 | 2 | 3")
        print("---+---+---")
        print(" 4 | 5 | 6")
        print("---+---+---")
        print(" 7 | 8 | 9")
        print()

# Test the display methods
board = Board()
board.display_reference()
board.display()

# Let's manually add some X's and O's to see how it looks
board.grid[0] = "X"  # Position 1
board.grid[4] = "O"  # Position 5
board.grid[8] = "X"  # Position 9

print("After adding some moves:")
board.display()
````

### 3.3 Move Validation - Ensuring Fair Play

````python
# Step 3: Add move validation and game state checking
class Board:
    def __init__(self):
        self.grid = [" "] * 9

    def display(self):
        print("\n")
        print(" " + self.grid[0] + " | " + self.grid[1] + " | " + self.grid[2])
        print("---+---+---")
        print(" " + self.grid[3] + " | " + self.grid[4] + " | " + self.grid[5])
        print("---+---+---")
        print(" " + self.grid[6] + " | " + self.grid[7] + " | " + self.grid[8])
        print()

    def is_full(self):
        """Check if the board is completely filled"""
        return " " not in self.grid

    def make_move(self, position, symbol):
        """Attempt to place a symbol at the given position"""
        # Convert from 1-9 (user input) to 0-8 (array index)
        index = position - 1
        
        # Validate position is in range
        if index < 0 or index > 8:
            print(f"❌ Invalid position {position}. Choose between 1-9.")
            return False
            
        # Check if position is already taken
        if self.grid[index] != " ":
            print(f"❌ Position {position} is already taken by '{self.grid[index]}'.")
            return False
            
        # Make the move
        self.grid[index] = symbol
        print(f"✅ Placed '{symbol}' at position {position}")
        return True

# Test move validation
board = Board()
print("=== Testing Move Validation ===")

# Test valid move
print("\n1. Testing valid move:")
success1 = board.make_move(5, "X")
board.display()

# Test invalid move (same spot)
print("2. Testing invalid move (same spot):")
success2 = board.make_move(5, "O")

# Test invalid position
print("\n3. Testing invalid position:")
success3 = board.make_move(10, "O")

# Test valid move in different position  
print("\n4. Testing another valid move:")
success4 = board.make_move(1, "O")
board.display()

print(f"\nResults: {success1}, {success2}, {success3}, {success4}")
````

### 3.4 Win Detection - The Smart Algorithm

<div class="info-box key-points">
    <h3>Win Detection Strategy:</h3>
    <ul>
        <li><strong>Data-driven:</strong> Store all winning combinations as data</li>
        <li><strong>Systematic:</strong> Check each combination methodically</li>
        <li><strong>Efficient:</strong> Stop as soon as we find a winner</li>
        <li><strong>Scalable:</strong> Easy to modify for different board sizes</li>
    </ul>
</div>

````python
# Step 4: Complete Board class with win detection
class Board:
    def __init__(self):
        self.grid = [" "] * 9

    def display(self):
        print("\n")
        print(" " + self.grid[0] + " | " + self.grid[1] + " | " + self.grid[2])
        print("---+---+---")
        print(" " + self.grid[3] + " | " + self.grid[4] + " | " + self.grid[5])
        print("---+---+---")
        print(" " + self.grid[6] + " | " + self.grid[7] + " | " + self.grid[8])
        print()

    def display_reference(self):
        print("\nPosition Reference:")
        print(" 1 | 2 | 3")
        print("---+---+---")
        print(" 4 | 5 | 6")
        print("---+---+---")
        print(" 7 | 8 | 9")
        print()

    def make_move(self, position, symbol):
        index = position - 1
        if 0 <= index <= 8 and self.grid[index] == " ":
            self.grid[index] = symbol
            return True
        return False

    def is_full(self):
        return " " not in self.grid

    def check_winner(self, symbol):
        """Check if the given symbol has won the game"""
        # All possible winning combinations (by index)
        win_combinations = [
            # Rows
            [0, 1, 2],  # Top row
            [3, 4, 5],  # Middle row  
            [6, 7, 8],  # Bottom row
            # Columns
            [0, 3, 6],  # Left column
            [1, 4, 7],  # Middle column
            [2, 5, 8],  # Right column
            # Diagonals
            [0, 4, 8],  # Top-left to bottom-right
            [2, 4, 6]   # Top-right to bottom-left
        ]
        
        # Check each winning combination
        for combo in win_combinations:
            if (self.grid[combo[0]] == symbol and
                self.grid[combo[1]] == symbol and
                self.grid[combo[2]] == symbol):
                return True
        return False

# Test win detection
print("=== Testing Win Detection ===")
board = Board()

# Create a winning scenario (top row)
print("Setting up a winning scenario...")
board.make_move(1, "X")  # Position 1
board.make_move(2, "X")  # Position 2
board.make_move(3, "X")  # Position 3 - should win!

board.display()

# Check for winner
if board.check_winner("X"):
    print("🎉 X wins with a top row!")
else:
    print("❌ No winner detected (this shouldn't happen!)")

# Test non-winner
print("\n=== Testing Non-Winner ===")
board2 = Board()
board2.make_move(1, "X")
board2.make_move(5, "O")
board2.make_move(9, "X")
board2.display()

if board2.check_winner("X"):
    print("🎉 X wins!")
else:
    print("❌ No winner yet - game continues")
````

<div class="section">
    <h2><span class="section-icon">4</span>Building Block 3: The TicTacToe Game Controller</h2>
    <p>The TicTacToe class is the "conductor" that orchestrates our entire game. It manages players, coordinates turns, and controls the game flow.</p>

    <div class="info-box key-points">
        <h3>Game Controller Responsibilities:</h3>
        <ul>
            <li><strong>Player Management:</strong> Keeps track of both players</li>
            <li><strong>Turn Control:</strong> Alternates between players</li>
            <li><strong>Game Flow:</strong> Manages the overall game loop</li>
            <li><strong>Coordination:</strong> Makes Player and Board classes work together</li>
        </ul>
    </div>

    <div class="info-box warning-box">
        <h4>🎯 Key OOP Principle: Composition</h4>
        <p>Notice how TicTacToe "has-a" Board and "has-a" list of Players. This is <strong>composition</strong> - building complex objects from simpler ones!</p>
    </div>
</div>

````python
# Complete Player class (for reference)
class Player:
    def __init__(self, name, symbol):
        self.name = name
        self.symbol = symbol

# Complete Board class (condensed for space)
class Board:
    def __init__(self):
        self.grid = [" "] * 9

    def display(self):
        print("\n")
        for i in range(0, 9, 3):
            print(f" {self.grid[i]} | {self.grid[i+1]} | {self.grid[i+2]}")
            if i < 6: print("---+---+---")
        print()

    def display_reference(self):
        print("\nPosition Reference:")
        for i in range(1, 10, 3):
            print(f" {i} | {i+1} | {i+2}")
            if i < 7: print("---+---+---")
        print()

    def make_move(self, position, symbol):
        index = position - 1
        if 0 <= index <= 8 and self.grid[index] == " ":
            self.grid[index] = symbol
            return True
        return False

    def is_full(self):
        return " " not in self.grid

    def check_winner(self, symbol):
        win_combinations = [
            [0,1,2], [3,4,5], [6,7,8],  # Rows
            [0,3,6], [1,4,7], [2,5,8],  # Columns  
            [0,4,8], [2,4,6]            # Diagonals
        ]
        return any(all(self.grid[pos] == symbol for pos in combo) for combo in win_combinations)

# Step 5: Create the TicTacToe game controller
class TicTacToe:
    def __init__(self, player1, player2):
        """Initialize the game with two players"""
        self.board = Board()              # Composition: Game "has-a" Board
        self.players = [player1, player2] # Store both players
        self.current_player = player1     # Start with player 1
        print(f"🎮 Game created! {player1.name} vs {player2.name}")
        print(f"📍 {self.current_player.name} goes first with '{self.current_player.symbol}'")

    def switch_player(self):
        """Switch to the other player"""
        # Toggle between players[0] and players[1]
        self.current_player = (
            self.players[1] if self.current_player == self.players[0] 
            else self.players[0]
        )
        print(f"🔄 Now it's {self.current_player.name}'s turn ({self.current_player.symbol})")

    def get_current_player_info(self):
        """Get information about the current player"""
        return f"{self.current_player.name} ({self.current_player.symbol})"

# Test the TicTacToe game controller
print("=== Testing Game Controller ===")
player1 = Player("Alice", "X")
player2 = Player("Bob", "O")
game = TicTacToe(player1, player2)

# Test player switching
print(f"\nStarting player: {game.get_current_player_info()}")
game.switch_player()
print(f"After switch: {game.get_current_player_info()}")
game.switch_player()
print(f"After another switch: {game.get_current_player_info()}")

# Show the empty board
print("\n=== Game Board ===")
game.board.display_reference()
game.board.display()
````

<div class="section">
    <h2><span class="section-icon">5</span>Putting It All Together - A Complete Game Demo</h2>
    <p>Now let's see all our classes working together in harmony! We'll simulate a complete game to demonstrate how the objects collaborate.</p>

    <div class="info-box key-points">
        <h3>Watch How the Classes Collaborate:</h3>
        <ul>
            <li><strong>TicTacToe</strong> orchestrates the overall game flow</li>
            <li><strong>Player</strong> objects provide name and symbol data</li>
            <li><strong>Board</strong> validates moves and detects wins</li>
            <li><strong>All together</strong> they create a seamless gaming experience</li>
        </ul>
    </div>
</div>

````python
# Complete game simulation
def simulate_game():
    print("🎯 === COMPLETE GAME SIMULATION === 🎯\n")
    
    # Create players
    alice = Player("Alice", "X")
    bob = Player("Bob", "O")
    
    # Create game
    game = TicTacToe(alice, bob)
    
    # Show reference board
    game.board.display_reference()
    
    # Simulate a series of moves
    moves = [
        (1, "Alice takes top-left corner"),
        (5, "Bob takes center"), 
        (2, "Alice takes top-middle"),
        (6, "Bob blocks Alice"),
        (3, "Alice completes top row - WINNER!")
    ]
    
    print("🎮 Starting Game:")
    game.board.display()
    
    # Play through each move
    for move_num, (position, description) in enumerate(moves, 1):
        print(f"\n--- Move {move_num}: {description} ---")
        
        # Attempt the move
        if game.board.make_move(position, game.current_player.symbol):
            # Show updated board
            game.board.display()
            
            # Check for winner
            if game.board.check_winner(game.current_player.symbol):
                print(f"🎉🎉🎉 {game.current_player.name} ({game.current_player.symbol}) WINS! 🎉🎉🎉")
                print(f"🏆 Victory achieved with symbol '{game.current_player.symbol}'!")
                break
                
            # Check for tie
            if game.board.is_full():
                print("🤝 It's a tie! Both players played well!")
                break
                
            # Switch to next player
            game.switch_player()
        else:
            print("❌ Move failed! (This shouldn't happen in our simulation)")

# Run the simulation
simulate_game()

print("\n" + "="*50)
print("🎯 Simulation complete! Notice how:")
print("   • Each class handled its own responsibilities")
print("   • Objects communicated through method calls")
print("   • The code remained clean and organized")
print("="*50)
````

````python
# Challenge Starter: Enhanced Player with Score Tracking
# Try completing this class!

class EnhancedPlayer(Player):
    def __init__(self, name, symbol):
        super().__init__(name, symbol)  # Call parent class constructor
        self.wins = 0
        self.games_played = 0
    
    def add_win(self):
        """Record a win for this player"""
        self.wins += 1
        print(f"🎉 {self.name} now has {self.wins} wins!")
    
    def add_game(self):
        """Record that this player played a game"""
        self.games_played += 1
    
    def get_win_rate(self):
        """Calculate and return the win percentage"""
        if self.games_played == 0:
            return 0
        return (self.wins / self.games_played) * 100
    
    def display_stats(self):
        """Show player statistics"""
        print(f"📊 {self.name} ({self.symbol}):")
        print(f"   Games: {self.games_played}")
        print(f"   Wins: {self.wins}")
        print(f"   Win Rate: {self.get_win_rate():.1f}%")

# Test the enhanced player
enhanced_player = EnhancedPlayer("Charlie", "Z")
enhanced_player.add_game()
enhanced_player.add_win()
enhanced_player.add_game()
enhanced_player.display_stats()

print("\n" + "="*40)
print("🎯 Your turn to experiment!")
print("Try adding more features to the classes above.")
print("Remember: each class should have a single, clear responsibility.")
print("="*40)
````
