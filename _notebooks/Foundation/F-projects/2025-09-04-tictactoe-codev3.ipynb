    <p>Let's start with the simplest component - a Player. In tic-tac-toe, each player needs a name and a symbol (X or O). The Player class will store this information.</p>

````python
class Player:
    def __init__(self, name, symbol):
        self.name = name
        self.symbol = symbol
````

Let's break down what just happened:
- `class Player:` creates a new class called Player
- `def __init__(self, name, symbol):` is the constructor - it runs when we create a new player
- `self.name = name` stores the player's name as an instance variable
- `self.symbol = symbol` stores the player's symbol (X or O)

Now let's test our Player class:

````python
# Let's create some players and see how it works
player1 = Player("Alice", "X")
player2 = Player("Bob", "O")

print(f"Player 1: {player1.name} uses symbol '{player1.symbol}'")
print(f"Player 2: {player2.name} uses symbol '{player2.symbol}'")
````

            <li><strong>self parameter:</strong> Refers to the specific player object being created</li>

### 3.1 Creating the Board Structure

First, we need to represent our 3x3 grid. We'll use a list with 9 elements, where each element represents one square:

````python
class Board:
    def __init__(self):
        self.grid = [" "] * 9  # Creates 9 empty spaces
        print("New board created!")
        print(f"Grid contents: {self.grid}")

# Test creating a board
test_board = Board()
````

Why 9 spaces? Our 3x3 board maps to list positions like this:

```
0 | 1 | 2
3 | 4 | 5  
6 | 7 | 8
```

When players enter positions 1-9, we'll convert them to indices 0-8.

<div class="info-box key-points">
    <h3>Board Initialization:</h3>
    <ul>
        <li><strong>self.grid = [" "] * 9:</strong> Creates a list with 9 empty spaces</li>
        <li><strong>Why 9 spaces?</strong> We represent the 3x3 grid as positions 0-8</li>
        <li><strong>Position mapping:</strong> User enters 1-9, we convert to 0-8 internally</li>
    </ul>

### 3.2 Adding Display Methods

Now let's add methods to display the board. We need two types of display:
1. Show the current game state
2. Show position numbers to help players

````python
class Board:
    def __init__(self):
        self.grid = [" "] * 9

    def display(self):
        print("\n")
        print(" " + self.grid[0] + " | " + self.grid[1] + " | " + self.grid[2])
        print("---+---+---")
        print(" " + self.grid[3] + " | " + self.grid[4] + " | " + self.grid[5])
        print("---+---+---")
        print(" " + self.grid[6] + " | " + self.grid[7] + " | " + self.grid[8])
        print("\n")

    def display_reference(self):
        reference = ["1", "2", "3", "4", "5", "6", "7", "8", "9"]
        print("Board positions:\n")
        print(" " + reference[0] + " | " + reference[1] + " | " + reference[2])
        print("---+---+---")
        print(" " + reference[3] + " | " + reference[4] + " | " + reference[5])
        print("---+---+---")
        print(" " + reference[6] + " | " + reference[7] + " | " + reference[8])
        print("\n")

# Test the display methods
board = Board()
print("This shows the position numbers:")
board.display_reference()
print("This shows the current game state:")
board.display()
````

Notice how each method has a single responsibility:
- `display()` shows the current game state
- `display_reference()` shows position numbers for player reference

<div class="info-box warning-box">
    <h4>🎯 Encapsulation in Action</h4>
    <p>The Board class knows how to display itself. We don't need external code to format the output - this is encapsulation!</p>

### 3.3 Adding Move Validation

Now let's add the ability to make moves and check if the board is full:

````python
class Board:
    def __init__(self):
        self.grid = [" "] * 9

    def display(self):
        print("\n")
        print(" " + self.grid[0] + " | " + self.grid[1] + " | " + self.grid[2])
        print("---+---+---")
        print(" " + self.grid[3] + " | " + self.grid[4] + " | " + self.grid[5])
        print("---+---+---")
        print(" " + self.grid[6] + " | " + self.grid[7] + " | " + self.grid[8])
        print("\n")

    def is_full(self):
        return " " not in self.grid

    def make_move(self, position, symbol):
        index = position - 1  # Convert 1-9 to 0-8
        if index < 0 or index > 8:
            print("Invalid position. Choose a number between 1 and 9.")
            return False
        if self.grid[index] != " ":
            print("That spot is already taken. Try again.")
            return False
        self.grid[index] = symbol
        return True
````

Let's break down the `make_move` method:
1. Convert user position (1-9) to array index (0-8)
2. Check if position is valid (between 0-8)
3. Check if the spot is already taken
4. If valid, place the symbol and return True
5. If invalid, show error message and return False

Now let's test the move validation:

````python
# Test the game logic
board = Board()
print("Testing valid move:")
result1 = board.make_move(5, "X")  # Should work
print(f"Move successful: {result1}")
board.display()

print("Testing invalid move (same spot):")
result2 = board.make_move(5, "O")  # Should fail
print(f"Move successful: {result2}")

print("Testing invalid position:")
result3 = board.make_move(10, "O")  # Should fail
print(f"Move successful: {result3}")
````

### 3.4 Win Detection - The Smart Algorithm

The most complex part of our Board class is checking for winners. Let's understand the strategy first:

<div class="info-box key-points">
    <h3>Win Detection Strategy:</h3>
    <ul>
        <li><strong>Data-driven:</strong> All winning combinations are stored as data, not hardcoded logic</li>
        <li><strong>Scalable:</strong> Easy to modify for different board sizes</li>
        <li><strong>Readable:</strong> The winning patterns are clearly visible</li>
        <li><strong>Efficient:</strong> Checks all possibilities in a simple loop</li>
    </ul>
</div>

````python
class Board:
    def __init__(self):
        self.grid = [" "] * 9

    def display(self):
        print("\n")
        print(" " + self.grid[0] + " | " + self.grid[1] + " | " + self.grid[2])
        print("---+---+---")
        print(" " + self.grid[3] + " | " + self.grid[4] + " | " + self.grid[5])
        print("---+---+---")
        print(" " + self.grid[6] + " | " + self.grid[7] + " | " + self.grid[8])
        print("\n")

    def make_move(self, position, symbol):
        index = position - 1
        if 0 <= index <= 8 and self.grid[index] == " ":
            self.grid[index] = symbol
            return True
        return False

    def is_full(self):
        return " " not in self.grid

    def check_winner(self, symbol):
        win_combinations = [
            [0, 1, 2], [3, 4, 5], [6, 7, 8],  # Rows
            [0, 3, 6], [1, 4, 7], [2, 5, 8],  # Columns
            [0, 4, 8], [2, 4, 6]              # Diagonals
        ]
        
        for combo in win_combinations:
            if (self.grid[combo[0]] == symbol and
                self.grid[combo[1]] == symbol and
                self.grid[combo[2]] == symbol):
                return True
        return False
````

Let's understand the win detection algorithm:

1. **win_combinations** lists all 8 ways to win:
   - 3 rows: [0,1,2], [3,4,5], [6,7,8]
   - 3 columns: [0,3,6], [1,4,7], [2,5,8] 
   - 2 diagonals: [0,4,8], [2,4,6]

2. For each combination, check if all three positions contain the same symbol

3. If any combination matches, we have a winner!

Let's test this:

````python
# Test win detection
board = Board()
print("Setting up a winning scenario...")
board.make_move(1, "X")  # Top left
board.make_move(2, "X")  # Top middle  
board.make_move(3, "X")  # Top right - should be a win!

board.display()
is_winner = board.check_winner("X")
print(f"Is X the winner? {is_winner}")
````

<div class="section">
    <h2><span class="section-icon">4</span>Step 3: The TicTacToe Game Controller</h2>
    <p>Now we need a class to bring everything together. The TicTacToe class will manage the players, the board, and control the game flow.</p>

    <div class="info-box warning-box">
        <h4>🎯 Key OOP Principle: Composition</h4>
        <p>TicTacToe "has-a" Board and "has-a" list of Players. This is called the Single Responsibility Principle - each class has one clear job!</p>
    </div>
</div>

````python
class TicTacToe:
    def __init__(self, player1, player2):
        self.board = Board()              # Composition: TicTacToe "has-a" Board
        self.players = [player1, player2] # Stores both players
        self.current_player = player1     # Tracks whose turn it is

    def switch_player(self):
        # Alternate between the two players
        self.current_player = (
            self.players[1] if self.current_player == self.players[0] else self.players[0]
        )
        print(f"Now it's {self.current_player.name}'s turn")
````

Let's break down what the TicTacToe class does:
- `self.board = Board()` creates a new board for this game
- `self.players = [player1, player2]` stores both players in a list
- `self.current_player = player1` keeps track of whose turn it is
- `switch_player()` alternates between the two players

Let's test the game controller:

````python
# Test the TicTacToe class setup
player1 = Player("Alice", "X")
player2 = Player("Bob", "O")
game = TicTacToe(player1, player2)

print(f"Game created! Current player: {game.current_player.name}")
game.switch_player()
game.switch_player()
````

    <h2><span class="section-icon">5</span>Putting It All Together - Complete Game Demo</h2>
    <p>Now let's see how all our classes work together! We'll create a complete working example where all the pieces collaborate seamlessly.</p>
    
    <div class="info-box key-points">
        <h3>Notice how the classes collaborate:</h3>
        <ul>
            <li><strong>TicTacToe</strong> manages the overall game flow</li>
            <li><strong>Board</strong> validates moves and checks for winners</li>
            <li><strong>Player</strong> provides the data needed for moves</li>

First, let's put together our complete classes:

````python
# Complete Player class
class Player:
    def __init__(self, name, symbol):
        self.name = name
        self.symbol = symbol

# Complete Board class
class Board:
    def __init__(self):
        self.grid = [" "] * 9

    def display(self):
        print("\n")
        print(" " + self.grid[0] + " | " + self.grid[1] + " | " + self.grid[2])
        print("---+---+---")
        print(" " + self.grid[3] + " | " + self.grid[4] + " | " + self.grid[5])
        print("---+---+---")
        print(" " + self.grid[6] + " | " + self.grid[7] + " | " + self.grid[8])
        print("\n")

    def display_reference(self):
        reference = ["1", "2", "3", "4", "5", "6", "7", "8", "9"]
        print("Board positions:\n")
        print(" " + reference[0] + " | " + reference[1] + " | " + reference[2])
        print("---+---+---")
        print(" " + reference[3] + " | " + reference[4] + " | " + reference[5])
        print("---+---+---")
        print(" " + reference[6] + " | " + reference[7] + " | " + reference[8])
        print("\n")

    def is_full(self):
        return " " not in self.grid

    def make_move(self, position, symbol):
        index = position - 1
        if index < 0 or index > 8:
            print("Invalid position. Choose a number between 1 and 9.")
            return False
        if self.grid[index] != " ":
            print("That spot is already taken. Try again.")
            return False
        self.grid[index] = symbol
        return True

    def check_winner(self, symbol):
        win_combinations = [
            [0, 1, 2], [3, 4, 5], [6, 7, 8],  # Rows
            [0, 3, 6], [1, 4, 7], [2, 5, 8],  # Columns
            [0, 4, 8], [2, 4, 6]              # Diagonals
        ]
        for combo in win_combinations:
            if (self.grid[combo[0]] == symbol and
                self.grid[combo[1]] == symbol and
                self.grid[combo[2]] == symbol):
                return True
        return False

# Complete TicTacToe class
class TicTacToe:
    def __init__(self, player1, player2):
        self.board = Board()
        self.players = [player1, player2]
        self.current_player = player1

    def switch_player(self):
        self.current_player = (
            self.players[1] if self.current_player == self.players[0] else self.players[0]
        )
        print(f"Now it's {self.current_player.name}'s turn")

print("All classes defined! Ready for the demo.")
````

Now let's simulate a complete game to see everything working together:

````python
# Let's simulate a quick game to see all classes working together
print("=== SIMULATED GAME DEMO ===")
player1 = Player("Alice", "X")
player2 = Player("Bob", "O")
game = TicTacToe(player1, player2)

# Simulate some moves
moves = [
    (1, "Alice plays position 1"),
    (5, "Bob plays position 5"), 
    (2, "Alice plays position 2"),
    (6, "Bob plays position 6"),
    (3, "Alice plays position 3 - should win!")
]

print(f"Starting game: {player1.name} vs {player2.name}")
game.board.display_reference()
game.board.display()

for position, description in moves:
    print(f"\n--- {description} ---")
    
    # Make the move
    success = game.board.make_move(position, game.current_player.symbol)
    
    if success:
        game.board.display()
        
        # Check for winner
        if game.board.check_winner(game.current_player.symbol):
            print(f"🎉 {game.current_player.name} ({game.current_player.symbol}) WINS!")
            break
            
        # Check for tie
        if game.board.is_full():
            print("It's a tie!")
            break
            
        # Switch players
        game.switch_player()
    else:
        print("Move failed, trying next move...")
````
